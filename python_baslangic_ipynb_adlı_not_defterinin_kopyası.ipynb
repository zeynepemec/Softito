### ✏️ Soru 24:
Bir alışveriş sepeti uygulaması yapın:
- `sepet = []` listesi oluşturun
- 3 ürün ekleyin (`.append()`)
- `for` döngüsü ile ürünleri numaralı yazdırın
- Kaç ürün olduğunu yazdırın

```python
# Buraya yazın:
sepet = []
sepet.append("Çanta")
sepet.append("Ayakkabı")
sepet.append("Elbise")

for sira, urun in enumerate(sepet, start=1):   #enumerate ile hem indexi hem ögeyi alırız
    print(f"{sira}. {urun}")

print(f"Sepetinizde toplam {len(sepet)} ürün bulunmaktadır.")   # sepetin uzunluğunu verir
```

### ✏️ Soru 25:
Bir listede verilen notların ortalamasını hesaplayan `ortalama(notlar)` fonksiyonu yazın.
- `sum()` ve `len()` kullanabilirsiniz
- `[70, 85, 90, 60, 75]` listesiyle test edin

```python
def ortalama(notlar):
    # Fonksiyonun içine girdik (4 boşluk içeriden başlıyoruz)
    toplam = sum(notlar)
    adet = len(notlar)
    sonuc = toplam / adet
    return sonuc

# Fonksiyon tanımı bitti, ana kod bloğu en soldan (başlangıçtan) devam etmeli
test_notlari = [70, 85, 90, 60, 75]

# print satırının başında asla boşluk olmamalı
print(ortalama(test_notlari))  # Çıktı: 76.0
```
