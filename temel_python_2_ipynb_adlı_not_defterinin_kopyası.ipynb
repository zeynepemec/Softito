`while` döngüsü ile basamak sayısı bulma: Verilen bir sayının kaç basamaklı olduğunu, sayıyı 10'a bölerek bulun (string dönüşümü kullanmadan).

```python
# Cevabınızı buraya yazın:
# Örnek: 4752 -> 4 basamaklı
sayi = 4752
sayi = int(input("Bir sayı girin: "))
gecici_sayi = abs(sayi) # Negatif sayılar için mutlak değer alıyoruz
sayac = 0

# Özel durum: Sayı 0 ise 1 basamaklıdır
if gecici_sayi == 0:
    sayac = 1
else:
    while gecici_sayi > 0:
        gecici_sayi //= 10  # Sayıyı 10'a böl (tam sayı bölmesi)
        sayac += 1          # Sayacı artır

print(f"Girdiğiniz sayı {sayac} basamaklıdır.")
```

### 🔸 Soru 12:
Bir öğrenci listesi var. Her öğrenci `(isim, not)` tuple'ı olarak saklanıyor.
- Notu 70'in üzerinde olanları filtreleyin
- En yüksek notu alan öğrenciyi bulun
- Notların ortalamasını hesaplayın

```python
# Cevabınızı buraya yazın:
ogrenciler = [
    ("Ahmet", 85),
    ("Büşra", 62),
    ("Can", 91),
    ("Deniz", 74),
    ("Elif", 58),
    ("Furkan", 88)
]

basarili_olanlar = [ogrenci for ogrenci in ogrenciler if ogrenci[1] > 70]

en_yuksek_ogrenci = max(ogrenciler, key=lambda x: x[1])

toplam_not = sum(n[1] for n in ogrenciler)
ortalama = toplam_not / len(ogrenciler)


print(f"Başarılı Öğrenciler: {basarili_olanlar}")
print(f"Zirvedeki İsim: {en_yuksek_ogrenci[0]} -> {en_yuksek_ogrenci[1]}")
print(f"Sınıf Ortalaması: {ortalama:.1f}")
```

### 🔸 Soru 13 (Set):
İki sınıfın öğrenci listesi var. `set` kullanarak:
- Her iki sınıfta da bulunan öğrencileri bulun (kesişim)
- Toplam kaç farklı öğrenci olduğunu bulun (birleşim)
- Sadece A sınıfında olup B'de olmayan öğrencileri bulun (fark)

```python
# Cevabınızı buraya yazın:
sinif_a = {"Ali", "Ayşe", "Can", "Dilan", "Ece"}
sinif_b = {"Can", "Dilan", "Fatih", "Gökçe", "Hakan"}

ortak_ogrenciler = sinif_a & sinif_b #kesişim

tum_ogrenciler = sinif_a | sinif_b #birleşim

sadece_a = sinif_a - sinif_b #fark

# 4. Ekstra: Simetrik Fark (Ya sadece A'da ya da sadece B'de olanlar) - '^' operatörü
sadece_bir_sinifta_olanlar = sinif_a ^ sinif_b

print(f"Kesişim (&): {ortak_ogrenciler}")
print(f"Toplam Öğrenci Sayısı (|): {len(tum_ogrenciler)}")
print(f"Sadece A'da olanlar (-): {sadece_a}")
print(f"Ortak olmayan herkes (^): {sadece_bir_sinifta_olanlar}")
```

### 🔸 Soru 15:
Özyinelemeli (recursive) bir fonksiyon yazın: `faktoriyel(n)` - n! hesaplasın.
- 0! = 1
- n! = n × (n-1)!

5!, 10! değerlerini hesaplayın.

```python
# Cevabınızı buraya yazın:

def faktoriyel(n):
    # Temel Durum (Base Case)
    if n == 0:
        return 1
    # Özyinelemeli Adım (Recursive Step)
    else:
        return n * faktoriyel(n - 1)

# Hesaplamalar
sonuc_5 = faktoriyel(5)
sonuc_10 = faktoriyel(10)

print(f"5! = {sonuc_5}")
print(f"10! = {sonuc_10}")
```

### 🔸 Soru 16:
`lambda` fonksiyonu kullanarak:
- Bir sayının karesini hesaplayan lambda
- İki sayının büyüğünü döndüren lambda
- `sorted()` ile bir isim listesini son harfe göre sıralayan lambda

```python
# Cevabınızı buraya yazın:
isimler = ["Zeynep", "Ali", "Mahmut", "Su", "Berk"]


# 1. Bir sayının karesini hesaplayan lambda
kare_al = lambda x: x**2

# 2. İki sayının büyüğünü döndüren lambda (Ternary operator ile)
buyuk_olan = lambda a, b: a if a > b else b

# 3. İsim listesini son harfe göre sıralayan lambda
# x[-1] listenin her bir elemanının son karakterini temsil eder
sirali_isimler = sorted(isimler, key=lambda x: x[-1])

# Sonuçları Yazdıralım
print(f"5'in karesi: {kare_al(5)}")
print(f"10 ve 25 arasından büyük olan: {buyuk_olan(10, 25)}")
print(f"Son harfe göre sıralı liste: {sirali_isimler}")
```

### 🔸 Soru 18 (Bütünleştirici):
Basit bir **öğrenci not sistemi** yazın:

- `ogrenciler` adında boş bir sözlük oluşturun (anahtar: isim, değer: notlar listesi)
- `ogrenci_ekle(isim)`: Yeni öğrenci ekler
- `not_ekle(isim, not_)`: Öğrenciye not ekler. Öğrenci yoksa hata mesajı verir.
- `ortalama_hesapla(isim)`: Öğrencinin not ortalamasını döndürür
- `sinif_raporu()`: Tüm öğrencileri ve ortalamalarını yazdırır, sınıf ortalamasını gösterir

En az 3 öğrenci ve her birine 3 not girerek test edin.

```python
# Cevabınızı buraya yazın:
ogrenciler = {}

def ogrenci_ekle(isim):
    if isim not in ogrenciler:
        ogrenciler[isim] = []
    else:
        print(f"Hata: {isim} zaten sistemde kayıtlı.")

def not_ekle(isim, not_):
    if isim in ogrenciler:
        ogrenciler[isim].append(not_)
    else:
        print(f"Hata: {isim} isminde bir öğrenci bulunamadı!")

def ortalama_hesapla(isim):
    notlar = ogrenciler.get(isim, [])
    if not notlar:
        return 0
    return sum(notlar) / len(notlar)

def sinif_raporu():
    print("\n--- SINIF RAPORU ---")
    toplam_ortalamalar = 0

    for isim in ogrenciler:
        ort = ortalama_hesapla(isim)
        toplam_ortalamalar += ort
        print(f"Öğrenci: {isim} | Notlar: {ogrenciler[isim]} | Ortalama: {ort:.2f}")

    genel_ortalama = toplam_ortalamalar / len(ogrenciler) if ogrenciler else 0
    print(f"\nSınıf Genel Ortalaması: {genel_ortalama:.2f}")
    print("--------------------\n")

# Test kodu:
ogrenci_ekle("Ahmet")
ogrenci_ekle("Büşra")
ogrenci_ekle("Can")

not_ekle("Ahmet", 85)
not_ekle("Ahmet", 90)
not_ekle("Ahmet", 78)

not_ekle("Büşra", 92)
not_ekle("Büşra", 88)
not_ekle("Büşra", 95)

not_ekle("Can", 70)
not_ekle("Can", 65)
not_ekle("Can", 73)

sinif_raporu()

```
