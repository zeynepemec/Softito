### 🔸 Soru 5:
`cumle = "  python programlama dili çok güçlüdür  "` değişkeni için:
- Baştaki ve sondaki boşlukları temizleyin (`.strip()`)
- İlk harfini büyük yapın (`.capitalize()`)
- Kaç kelimeden oluştuğunu bulun (`.split()` ve `len()` kullanın)
- "güçlü" kelimesinin cümlede olup olmadığını kontrol edin (`in` operatörü)

```python
# Cevabınızı buraya yazın:
cumle = " python programlama dili çok güçlüdür "

temiz_cumle = cumle.strip() #boşlukları siler
buyuk_harf = temiz_cumle.capitalize() #ilk harf büyük
kelimeler = temiz_cumle.split() #ayırır
kelime_sayisi = len(kelimeler) #kaç kelimeden oluştuğunu bulur
var_mi = "güçlü" in temiz_cumle #güçlü kelimesi cümlede var mı diye bakar

print(f"Temizlenmiş Hali: '{temiz_cumle}'")
print(f"İlk Harf Büyük: {buyuk_harf}")
print(f"Kelime Listesi: {kelimeler}")
print(f"Kelime Sayısı: {kelime_sayisi}")
print(f"'güçlü' kelimesi var mı?: {var_mi}")

```

### 🔸 Soru 12:

### 🔸 Soru 13 (List Comprehension):
List comprehension kullanarak:
- 1-20 arasındaki çift sayıların listesini oluşturun
- Bir kelime listesindeki 4'ten uzun kelimeleri filtreleyin

```python
# Cevabınızı buraya yazın:
kelimeler = ["at", "kaplan", "it", "fil", "kedi", "aslan", "köpek"]

cift_sayilar = [sayi for sayi in range(1, 21) if sayi % 2 == 0]

kelimeler = ["at", "kaplan", "it", "fil", "kedi", "aslan", "köpek"]
uzun_kelimeler = [kelime for kelime in kelimeler if len(kelime) > 4]


print(f"Çift Sayılar: {cift_sayilar}")
print(f"4'ten Uzun Kelimeler: {uzun_kelimeler}")
```

### 🔸 Soru 15:
Bir sayının asal olup olmadığını kontrol eden `asal_mi(n)` fonksiyonu yazın.
- 1'den 20'ye kadar olan asal sayıları bu fonksiyonu kullanarak yazdırın.

```python
#Bir sayının asal olup olmadığını kontrol eden asal_mi(n) fonksiyonu yazın.

#1'den 20'ye kadar olan asal sayıları bu fonksiyonu kullanarak yazdırın.
def asal_mi(n):

    if n <= 1:
        return False


    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True

print("1-20 arasındaki asal sayılar:")
for sayi in range(1, 21):
    if asal_mi(sayi):
        print(sayi, end=" ")
```

### 🔸 Soru 17:
Kullanıcıdan bir sayı alındığını varsayın (`girdi = "abc"`).
Bu değeri int'e çevirmeye çalışırken `ValueError` hatası oluşabilir.
try-except ile bunu yakalayan ve anlamlı bir mesaj veren kod yazın.

```python
# Cevabınızı buraya yazın:
girdi = "abc"

try:

    sayi = int(girdi)
    print(f"Dönüştürme başarılı: {sayi}")
except ValueError:

    print("Hata: Girdiğiniz değer bir tam sayı değil! Lütfen sadece rakam kullanın.")
except Exception as e:

    print(f"Beklenmedik bir hata oluştu: {e}")
```

### 🔸 Soru 18 (Bütünleştirici):
Bir **not defteri uygulaması** yazın:
- `notlar` adında boş bir liste oluşturun
- `not_ekle(metin)` fonksiyonu: Notu listeye ekler
- `not_listele()` fonksiyonu: Tüm notları numaralandırarak yazdırır
- `not_sil(index)` fonksiyonu: Verilen indexteki notu siler, hatalı index için hata mesajı verir

Fonksiyonları test edin: 3 not ekleyin, listeleyin, ortadakini silin, tekrar listeleyin.

```python
# Cevabınızı buraya yazın:
notlar = []

#def (define) fonksiyon tanımlamak için kullanılır

def not_ekle(metin): # notu listeye ekle
    notlar.append(metin) #sona ekler

def not_listele(): #tüm notları numaralandırarak yazdırır.
    for i, n in enumerate(notlar, 1):
        print(f"{i}. {n}")

def not_sil(index):
    if 0 < index <= len(notlar):
        print(f"Silindi: {notlar.pop(index - 1)}")
    else:
        print("Geçersiz index!")

# Test
not_ekle("Python öğren")
not_ekle("Alıştırmaları tamamla")
not_ekle("Proje yap")
not_listele()
not_sil(2)
not_listele()
```
