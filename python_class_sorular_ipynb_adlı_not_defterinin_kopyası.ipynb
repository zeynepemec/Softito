### Soru 1 — Temel class
Aşağıdaki özelliklere sahip bir `Kisi` class'ı yaz:
- `ad`, `soyad`, `yas` instance değişkenleri olsun
- `tanitim()` metodu `"Merhaba, ben Ali Veli, 25 yaşındayım."` formatında yazdırsın

```python
# Cevabını buraya yaz
class Kisi:
    def __init__(self, ad, soyad, yas):
        # Bilgileri kaydediyoruz
        self.ad = ad
        self.soyad = soyad
        self.yas = yas

    def tanitim(self):
        # Kaydettiğimiz bilgileri yazdırıyoruz
        print(f"Merhaba, ben {self.ad} {self.soyad}, {self.yas} yaşındayım.")

# Kullanımı:
birisi = Kisi("Ali", "Veli", 25)
birisi.tanitim()
```

### Soru 2 — Metotlar
Bir `Dikdortgen` class'ı yaz:
- `en` ve `boy` alsın
- `alan()` → en × boy döndürsün
- `cevre()` → 2 × (en + boy) döndürsün
- `kare_mi()` → en == boy ise `True` döndürsün

```python
# Cevabını buraya yaz
class Dikdortgen:
    def __init__(self, en, boy):
        self.en = en
        self.boy = boy

    def alan(self):
        return self.en * self.boy

    def cevre(self):
        return 2 * (self.en + self.boy)

    def kare_mi(self):
        return self.en == self.boy

# Kullanımı:
dikdortgen = Dikdortgen(3, 5)

print(f"Alan: {dikdortgen.alan()}")
print(f"Çevre: {dikdortgen.cevre()}")
print(f"Kare mi?: {dikdortgen.kare_mi()}")

```

```python
# TEST
d1 = Dikdortgen(5, 3)
d2 = Dikdortgen(4, 4)
assert d1.alan() == 15
assert d1.cevre() == 16
assert d1.kare_mi() == False
assert d2.kare_mi() == True
print("✅ Soru 2 doğru!")
```

### Soru 3 — `__str__` ve `__repr__`
Bir `Kitap` class'ı yaz:
- `ad` ve `yazar` alsın
- `__str__`  → `"Suç ve Ceza — Dostoyevski"` formatında döndürsün
- `__repr__` → `"Kitap('Suç ve Ceza', 'Dostoyevski')"` formatında döndürsün

```python
# Cevabını buraya yaz
class Kitap:
    def __init__(self, ad, yazar):
        self.ad = ad
        self.yazar = yazar

    # Kullanıcı için (print(kitap) dendiğinde çalışır)
    def __str__(self):
        return f"{self.ad} — {self.yazar}"

    # Yazılımcı için (Nesnenin nasıl oluşturulduğunu gösterir)
    def __repr__(self):
        return f"Kitap('{self.ad}', '{self.yazar}')"

# Test edelim:
kitap1 = Kitap("Suç ve Ceza", "Dostoyevski")

print(str(kitap1))   # Çıktı: Suç ve Ceza — Dostoyevski
print(repr(kitap1))  # Çıktı: Kitap('Suç ve Ceza', 'Dostoyevski')
```

```python
# TEST
k = Kitap("Suç ve Ceza", "Dostoyevski")
assert str(k) == "Suç ve Ceza — Dostoyevski"
assert repr(k) == "Kitap('Suç ve Ceza', 'Dostoyevski')"
print("✅ Soru 3 doğru!")
```

### Soru 5 — Kalıtım
`Hayvan` class'ından türeyen `Kopek` ve `Kedi` class'ları yaz:
- `Hayvan`: `isim` alsın, `ses_cikar()` → `"..."` döndürsün
- `Kopek`: `ses_cikar()` → `"Hav!"` döndürsün, `getir()` → `"[isim] topu getirdi!"` yazdırsın
- `Kedi`: `ses_cikar()` → `"Miyav!"` döndürsün

```python
# Cevabını buraya yaz
class Hayvan:
    def __init__(self, isim):
        self.isim = isim

    def ses_cikar(self):
        return "..."

class Kopek(Hayvan): # Hayvan sınıfından türüyor
    def ses_cikar(self):
        return "Hav!"

    def getir(self):
        print(f"{self.isim} topu getirdi!")

class Kedi(Hayvan): # Hayvan sınıfından türüyor
    def ses_cikar(self):
        return "Miyav!"

# Test edelim:
karabas = Kopek("Karabaş")
pamuk = Kedi("Pamuk")

print(f"{karabas.isim}: {karabas.ses_cikar()}")
karabas.getir()

print(f"{pamuk.isim}: {pamuk.ses_cikar()}")
```

```python
# TEST
k = Kopek("Karabaş")
c = Kedi("Pamuk")
assert k.isim == "Karabaş"
assert k.ses_cikar() == "Hav!"
assert c.ses_cikar() == "Miyav!"
assert isinstance(k, Hayvan)
assert isinstance(c, Hayvan)
k.getir()  # Karabaş topu getirdi!
print("✅ Soru 5 doğru!")
```

### Soru 6 — Kapsülleme ve @property
Bir `Sicaklik` class'ı yaz:
- `_derece` ile saklansın
- `@property` ile `derece` okunabilsin
- `@derece.setter` ile atanabilsin ama `-273`'ün altına düşürülemesin (mutlak sıfır), düşürülmeye çalışılınca `ValueError` fırlatsın
- `@property` ile `fahrenheit` hesaplansın → `(derece × 9/5) + 32`

```python
# Cevabını buraya yaz
class Sicaklik:
    def __init__(self, derece):
        # Setter metodunu tetiklemek için doğrudan atama yapıyoruz
        self.derece = derece

    @property
    def derece(self):

        return self._derece

    @derece.setter
    def derece(self, deger):

        if deger < -273:
            raise ValueError("Sıcaklık mutlak sıfırın (-273°C) altına düşemez!")
        self._derece = deger

    @property
    def fahrenheit(self):
        """Dereceyi Fahrenheit cinsinden hesaplar."""
        return (self._derece * 9/5) + 32

# Test edelim:
try:
    hava = Sicaklik(25)
    print(f"Derece: {hava.derece}°C")
    print(f"Fahrenheit: {hava.fahrenheit}°F")

    # Hata fırlatacak deneme:
    hava.derece = -300
except ValueError as e:
    print(f"Hata yakalandı: {e}")
```

```python
# TEST
s = Sicaklik(100)
assert s.derece == 100
assert s.fahrenheit == 212.0
s.derece = 0
assert s.fahrenheit == 32.0
try:
    s.derece = -300
    assert False, "ValueError bekleniyor"
except ValueError:
    pass
print("✅ Soru 6 doğru!")
```

### Soru 7 — Dunder metotlar
Bir `Para` class'ı yaz:
- `miktar` ve `para_birimi` alsın (varsayılan: `"TL"`)
- `__str__`  → `"100 TL"` formatında
- `__add__`  → iki Para toplanabilsin (aynı para birimi kontrolü yap, farklıysa `ValueError`)
- `__eq__`   → miktar ve para birimi eşitse `True`
- `__lt__`   → miktar karşılaştırması

```python
# Cevabını buraya yaz
class Para:
    def __init__(self, miktar, para_birimi="TL"):
        self.miktar = miktar
        self.para_birimi = para_birimi

    def __str__(self):
        # "100 TL" formatında çıktı verir
        return f"{self.miktar} {self.para_birimi}"

    def __add__(self, diger):
        # Para birimleri aynı mı kontrol et
        if self.para_birimi != diger.para_birimi:
            raise ValueError("Farklı para birimleri arasında toplama yapılamaz!")
        # Yeni bir Para nesnesi döndür
        return Para(self.miktar + diger.miktar, self.para_birimi)

    def __eq__(self, diger):
        # Hem miktar hem de birim eşit olmalı
        return self.miktar == diger.miktar and self.para_birimi == diger.para_birimi

    def __lt__(self, diger):
        # Küçük mü karşılaştırması (Less Than)
        if self.para_birimi != diger.para_birimi:
            raise ValueError("Farklı para birimleri karşılaştırılamaz!")
        return self.miktar < diger.miktar

# Test edelim:
cüzdan1 = Para(100, "TL")
cüzdan2 = Para(50, "TL")
dolar = Para(10, "USD")

print(cüzdan1)           # 100 TL
print(cüzdan1 + cüzdan2) # 150 TL
print(cüzdan1 == cüzdan2) # False
print(cüzdan2 < cüzdan1)  # True
```

```python
# TEST
p1 = Para(100, "TL")
p2 = Para(50, "TL")
p3 = Para(100, "TL")
p4 = Para(50, "USD")
assert str(p1) == "100 TL"
assert (p1 + p2).miktar == 150
assert p1 == p3
assert p2 < p1
try:
    p1 + p4
    assert False, "ValueError bekleniyor"
except ValueError:
    pass
print("✅ Soru 7 doğru!")
```

### Soru 8 — @classmethod ve @staticmethod
Bir `Dikdortgen` class'ı yaz:
- `en` ve `boy` alsın
- `@classmethod kare_olustur(cls, kenar)` → kare döndürsün (en == boy == kenar)
- `@staticmethod gecerli_mi(en, boy)` → ikisi de 0'dan büyükse `True` döndürsün

```python
# Cevabını buraya yaz
class Dikdortgen:
    def __init__(self, en, boy):
        self.en = en
        self.boy = boy

    @classmethod
    def kare_olustur(cls, kenar):
        """Bir kenar uzunluğu alarak eni ve boyu eşit bir Dikdortgen (kare) döndürür."""
        return cls(kenar, kenar)

    @staticmethod
    def gecerli_mi(en, boy):
        """En ve boy değerlerinin geometrik olarak uygun olup olmadığını kontrol eder."""
        return en > 0 and boy > 0

# Test edelim:
# 1. Normal Dikdörtgen oluşturma
dik = Dikdortgen(5, 10)

# 2. Class Method ile Kare oluşturma
kare = Dikdortgen.kare_olustur(7)
print(f"Kare boyutları: {kare.en}x{kare.boy}")

# 3. Static Method ile kontrol (Nesne oluşturmadan çağrılabilir)
print(f"3 ve 5 geçerli mi?: {Dikdortgen.gecerli_mi(3, 5)}")
print(f"-2 ve 4 geçerli mi?: {Dikdortgen.gecerli_mi(-2, 4)}")
```

```python
# TEST
d = Dikdortgen(5, 3)
kare = Dikdortgen.kare_olustur(4)
assert kare.en == 4
assert kare.boy == 4
assert Dikdortgen.gecerli_mi(5, 3) == True
assert Dikdortgen.gecerli_mi(-1, 3) == False
assert Dikdortgen.gecerli_mi(0, 3) == False
print("✅ Soru 8 doğru!")
```

### Soru 9 — Iterator class
Bir `AralıkIterator` class'ı yaz:
- `baslangic`, `bitis`, `adim` alsın

- Python'un `range()` gibi çalışsın ama class olarak
- `__len__` da ekle — kaç eleman üreteceğini döndürsün

```python
# Cevabını buraya yaz
class AralikIterator:
    def __init__(self, baslangic, bitis, adim=1):
        self.baslangic = baslangic
        self.bitis = bitis
        self.adim = adim
        self.su_an = baslangic

    def __iter__(self):
        # Nesnenin kendisini bir iterator olarak döndürür
        return self

    def __next__(self):
        # Bitiş değerine ulaşıldı mı kontrol et
        if (self.adim > 0 and self.su_an >= self.bitis) or \
           (self.adim < 0 and self.su_an <= self.bitis):
            raise StopIteration

        # Mevcut değeri sakla, bir sonraki adımı hazırla ve döndür
        deger = self.su_an
        self.su_an += self.adim
        return deger

    def __len__(self):
        # Üretilecek toplam eleman sayısını hesaplar
        if (self.adim > 0 and self.baslangic >= self.bitis) or \
           (self.adim < 0 and self.baslangic <= self.bitis):
            return 0
        return abs((self.bitis - self.baslangic + self.adim - (1 if self.adim > 0 else -1)) // self.adim)

       ## def __len__(self):
## import math
## return max(0, math.ceil((self.bitis - self.baslangic) / self.adim))

# Test edelim:
sayici = AralikIterator(0, 10, 2)

print(f"Toplam eleman sayısı: {len(sayici)}")

for sayi in sayici:
    print(sayi)
```

### Soru 10 — Mixin
Aşağıdaki iki Mixin'i yaz ve `Calisan` class'ına ekle:
- `KarsilastirmaMixin`: `__eq__` ve `__lt__` eklesin — karşılaştırma `maas` değerine göre yapılsın
- `YazdirmaMixin`: `yazdir()` metodu tüm instance değişkenlerini `anahtar: değer` formatında yazdırsın
- `Calisan`: `isim`, `departman`, `maas` alsın

```python
# Cevabını buraya yaz
class KarsilastirmaMixin:
    """Maaş değerine göre karşılaştırma yeteneği kazandırır."""
    def __eq__(self, diger):
        return self.maas == diger.maas

    def __lt__(self, diger):
        return self.maas < diger.maas

class YazdirmaMixin:
    """Nesne değişkenlerini anahtar: değer formatında yazdırır."""
    def yazdir(self):
        # vars(self) veya self.__dict__ nesnenin tüm değişkenlerini getirir
        for anahtar, deger in vars(self).items():
            print(f"{anahtar}: {deger}")

class Calisan(KarsilastirmaMixin, YazdirmaMixin):
    def __init__(self, isim, departman, maas):
        self.isim = isim
        self.departman = departman
        self.maas = maas

# Test edelim:
calisan1 = Calisan("Zeynep", "Yazılım", 50000)
calisan2 = Calisan("Ali", "Tasarım", 45000)

# Yazdırma Mixin testi
print("--- Çalışan Bilgileri ---")
calisan1.yazdir()

# Karşılaştırma Mixin testi
print(f"\n{calisan1.isim} daha mı çok kazanıyor? {calisan2 < calisan1}")
```

```python
# TEST
c1 = Calisan("Ali", "Yazılım", 50000)
c2 = Calisan("Ayşe", "Tasarım", 60000)
c3 = Calisan("Ali", "Yazılım", 50000)
assert c1 < c2
assert c2 > c1
assert c1 == c3
c1.yazdir()
# isim: Ali
# departman: Yazılım
# maas: 50000
print("✅ Soru 10 doğru!")
```

### Soru 11 — Hepsini Birleştir (Proje)
Bir `Kutuphane` sistemi yaz:

**`Kitap` class'ı:**
- `isbn`, `ad`, `yazar`, `mevcut` (varsayılan `True`) alsın
- `__str__` → `"[ISBN] Ad — Yazar"`
- `__repr__` → `"Kitap(isbn, ad, yazar)"`

**`Kutuphane` class'ı:**
- `kitaplar` listesi tut
- `kitap_ekle(kitap)` → listeye ekle
- `__len__` → toplam kitap sayısı
- `__contains__` → ISBN ile kitap var mı kontrolü (`"978..." in kutuphane`)
- `kitap_bul(isbn)` → ISBN ile kitap döndür, bulamazsa `None`
- `odunc_al(isbn)` → kitabı `mevcut=False` yap, zaten alınmışsa hata ver
- `iade_et(isbn)` → kitabı `mevcut=True` yap
- `mevcut_kitaplar()` → sadece mevcut kitapları listele

```python
# Cevabını buraya yaz
class Kitap:
    def __init__(self, ad, yazar, isbn):
        self.ad = ad
        self.yazar = yazar
        self.isbn = isbn
        self.mevcut = True  # Kitap ilk eklendiğinde kütüphanededir

    def __repr__(self):
        return f"Kitap('{self.ad}', '{self.isbn}', Mevcut: {self.mevcut})"

class Kutuphane:
    def __init__(self):
        self.kitaplar = []

    def kitap_ekle(self, kitap):
        """Listeye yeni bir kitap nesnesi ekler."""
        self.kitaplar.append(kitap)

    def __len__(self):
        """len(kutuphane) dendiğinde toplam kitap sayısını döndürür."""
        return len(self.kitaplar)

    def __contains__(self, isbn):
        """'ISBN' in kutuphane dendiğinde varlık kontrolü yapar."""
        return any(k.isbn == isbn for k in self.kitaplar)

    def kitap_bul(self, isbn):
        """ISBN ile kitabı döndürür, bulamazsa None döndürür."""
        for kitap in self.kitaplar:
            if kitap.isbn == isbn:
                return kitap
        return None

    def odunc_al(self, isbn):
        """Kitabı ödünç verir (mevcut=False)."""
        kitap = self.kitap_bul(isbn)
        if not kitap:
            raise ValueError("Kitap kütüphanede kayıtlı değil!")
        if not kitap.mevcut:
            raise ValueError("Kitap zaten ödünç alınmış!")

        kitap.mevcut = False
        print(f"{kitap.ad} ödünç verildi.")

    def iade_et(self, isbn):
        """Kitabı geri alır (mevcut=True)."""
        kitap = self.kitap_bul(isbn)
        if kitap:
            kitap.mevcut = True
            print(f"{kitap.ad} iade edildi.")

    def mevcut_kitaplar(self):
        """Sadece kütüphanede olan (mevcut=True) kitapları döndürür."""
        return [k for k in self.kitaplar if k.mevcut]

# --- Kullanım Örneği ---
kutuphane = Kutuphane()

# Kitapları oluştur ve ekle
k1 = Kitap("Suç ve Ceza", "Dostoyevski", "97801")
k2 = Kitap("Nutuk", "Mustafa Kemal Atatürk", "97802")

kutuphane.kitap_ekle(k1)
kutuphane.kitap_ekle(k2)

# Testler
print(f"Toplam kitap: {len(kutuphane)}")
print(f"97801 var mı?: {'97801' in kutuphane}")

kutuphane.odunc_al("97801")
print(f"Mevcut kitaplar: {kutuphane.mevcut_kitaplar()}")

kutuphane.iade_et("97801")
```

```python
# TEST
k1 = Kitap("978-1", "Suç ve Ceza", "Dostoyevski")
k2 = Kitap("978-2", "1984", "Orwell")
k3 = Kitap("978-3", "Dune", "Herbert")

kutuphane = Kutuphane()
kutuphane.kitap_ekle(k1)
kutuphane.kitap_ekle(k2)
kutuphane.kitap_ekle(k3)

assert len(kutuphane) == 3
assert "978-1" in kutuphane
assert "978-9" not in kutuphane

kutuphane.odunc_al("978-1")
assert k1.mevcut == False
assert len(kutuphane.mevcut_kitaplar()) == 2

kutuphane.iade_et("978-1")
assert k1.mevcut == True

try:
    kutuphane.odunc_al("978-1")
    kutuphane.odunc_al("978-1")  # zaten alınmış
    assert False
except Exception:
    pass

print("✅ Soru 11 doğru! Tebrikler!")
```
